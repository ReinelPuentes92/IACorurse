## <span style="color:orange">**Simple Prompts | Templates = Plantilla o Estructura**</span>

```python
from langchain.chat_models import init_chat_model

# LangChain v1.0: init_chat_model detecta automáticamente el proveedor
chat_model = init_chat_model("gpt-4.1")
```

```python
# Los prompts se importan de langchain_core (no están en el namespace simplificado de v1.0)
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template('''
Responda la siguiente pregunta del usuario:
{pregunta_k}
''')

prompt_template
```

```python
print(prompt_template.format(pregunta_k='¿Qué es un agujero negro?'))
```

```python
prompt_template_2 = PromptTemplate.from_template('''¿Responda la siguiente pregunta del usuario con hasta {n_palabras} palabras:{pregunta}''')

prompt_template_2.format(n_palabras=10, pregunta='¿Qué es un agujero negro?')
```

```python
print(prompt_template_2.format(n_palabras=10, pregunta='¿Qué es un agujero negro?'))
```

```python
prompt_template_3 = PromptTemplate.from_template('''Responda la siguiente pregunta del usuario con hasta {n_palabras} palabras:{pregunta}''', partial_variables={'n_palabras': 10})

prompt_template_3.format(pregunta='¿Qué es un agujero negro?')
```

```python
print(prompt_template_3.format(pregunta='¿Qué es un agujero negro?'))
```

## <span style="color:orange">**Composing Prompts | Uniendo múltiples prompts**</span>

```python
# Los prompts se importan de langchain_core
from langchain_core.prompts import PromptTemplate

template_word_count = PromptTemplate.from_template('''Responda la pregunta con hasta {n_palabras} palabras.''')
template_idioma = PromptTemplate.from_template('''Retorna la respuesta en {idioma}.''')
template_final = (
    template_word_count
    + '\n'
    + template_idioma
    + '\nResponda la siguiente pregunta: {pregunta}'
)
template_final.template
```

```python
prompt = template_final.format(n_palabras=10, idioma='inglés', pregunta='¿Qué es una estrella?')
print(prompt)
```

```python
prompt = template_final.format(n_palabras=10, idioma='inglés', pregunta='¿Qué es una estrella?')
chat_model.invoke(prompt)
```

## <span style="color:orange">**Templates para Chat**</span>

```python
# Los prompts se importan de langchain_core
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_template('Esta es mi duda: {duda}')
chat_template.format_messages(duda='¿Quién soy yo?')
```

```python
chat_template_2 = ChatPromptTemplate.from_messages(
    [
        ('system', 'Eres un asistente gracioso y te llamas {nombre_asistente}'),
        ('human', 'Hola, cómo vas?'),
        ('ai', 'Mejor ahora! como puedo ayudarte?'),
        ('human', '{pregunta}')
    ]
)

chat_template_2
```

```python
chat_template_2.format_messages(nombre_asistente='Databot', pregunta='Cuál es tu nombre?')
```

```python
from langchain.chat_models import init_chat_model

# LangChain v1.0: init_chat_model detecta automáticamente el proveedor
chat_model = init_chat_model("gpt-4.1")

chat_model.invoke(chat_template_2.format_messages(nombre_asistente='Databot', pregunta='¿Cuál es tu nombre?'))
```

## <span style="color:orange">**Templates de Few-shot prompting para llm**</span>

```python
ejemplos = [
    {"pregunta": "¿Quién vivió más tiempo, Muhammad Ali o Alan Turing?", 
     "respuesta": 
     """¿Son necesarias preguntas de seguimiento aquí? Sí. 
Pregunta de seguimiento: ¿Cuántos años tenía Muhammad Ali cuando murió? 
Respuesta intermedia: Muhammad Ali tenía 74 años cuando murió.
Pregunta de seguimiento: ¿Cuántos años tenía Alan Turing cuando murió?
Respuesta intermedia: Alan Turing tenía 41 años cuando murió.
Entonces, la respuesta final es: Muhammad Ali.
""", 
    }, 
    {"pregunta": "¿Cuándo nació el fundador de Craigslist?", 
     "respuesta": 
"""¿Son necesarias preguntas de seguimiento aquí? Sí.
Pregunta de seguimiento: ¿Quién fue el fundador de Craigslist?
Respuesta intermedia: Craigslist fue fundado por Craig Newmark.
Pregunta de seguimiento: ¿Cuándo nació Craig Newmark?
Respuesta intermedia: Craig Newmark nació el 6 de diciembre de 1952.
Entonces, la respuesta final es: 6 de diciembre de 1952.
""", 
    }, 
    {"pregunta": "¿Quién fue el abuelo materno de George Washington?",
     "respuesta":
"""¿Son necesarias preguntas de seguimiento aquí? Sí.
Pregunta de seguimiento: ¿Quién fue la madre de George Washington?
Respuesta intermedia: La madre de George Washington fue Mary Ball Washington.
Pregunta de seguimiento: ¿Quién fue el padre de Mary Ball Washington?
Respuesta intermedia: El padre de Mary Ball Washington fue Joseph Ball.
Entonces, la respuesta final es: Joseph Ball
""",
    },
    {"pregunta": "¿Los directores de Jaws y Casino Royale son del mismo país?",
    "respuesta":
"""¿Son necesarias preguntas de seguimiento aquí? Sí.
Pregunta de seguimiento: ¿Quién es el director de Jaws?
Respuesta intermedia: El director de Jaws es Steven Spielberg.
Pregunta de seguimiento: ¿De dónde es Steven Spielberg?
Respuesta intermedia: Estados Unidos.
Pregunta de seguimiento: ¿Quién es el director de Casino Royale?
Respuesta intermedia: El director de Casino Royale es Martin Campbell.
Pregunta de seguimiento: ¿De dónde es Martin Campbell?
Respuesta intermedia: Nueva Zelanda.
Entonces, la respuesta final es: No
""",
    },
]
```

```python
# Los prompts se importan de langchain_core
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate

example_prompt = PromptTemplate(
    input_variables=['pregunta', 'respuesta'],
    template='Pregunta {pregunta}\n{respuesta}'
)

example_prompt.format(**ejemplos[0]) #Con 0 puedo ver mi primer ejemplo, con el 1 puedo ver mi segundo ejemplo
```

```python
prompt = FewShotPromptTemplate(
    examples=ejemplos,
    example_prompt=example_prompt,
    suffix='Pregunta: {input}',
    input_variables=['input']
)
prompt
```

```python
print(prompt.format(input='¿Quién hizo más goles, Romário o Pelé?'))
```

```python
chat_model.invoke(prompt.format(input='¿Quién hizo más goles, Romário o Pelé?'))
```

## <span style="color:orange">**Templates de Few-shot prompting para chat**</span>

```python
from langchain.chat_models import init_chat_model

# LangChain v1.0: init_chat_model detecta automáticamente el proveedor
chat_model = init_chat_model("gpt-4.1")
```

```python
# Los prompts se importan de langchain_core
from langchain_core.prompts import FewShotChatMessagePromptTemplate, ChatPromptTemplate

example_prompt = ChatPromptTemplate.from_messages(
    [('human', '{pregunta}'),
     ('ai', '{respuesta}')]
)

print(example_prompt.format_messages(**ejemplos[0]))
```

```python
few_shot_template = FewShotChatMessagePromptTemplate(
    examples=ejemplos,
    example_prompt=example_prompt
)

prompt_template = ChatPromptTemplate.from_messages(
    [
    few_shot_template,
    ('human', '{input}')
    ]
)

prompt = prompt_template.format_messages(input='¿Quién hizo más goles, Romário o Pelé?')
prompt
```

```python
chat_model.invoke(prompt)
```
