### Prompt few-shot

````python
from langchain.chat_models import init_chat_model

# LangChain v1.0: init_chat_model detecta automáticamente el proveedor
chat_model = init_chat_model("gpt-4.1")
````

````python
# LangChain v1.0: Messages se importan del namespace simplificado
from langchain.messages import SystemMessage, HumanMessage, AIMessage

mensajes = [
    SystemMessage(content='Eres un asistente que resuelve operaciones matemáticas y cuya respuesta es directa.'),
    HumanMessage(content='¿Cuánto es 1 + 1?'),
    AIMessage(content='2'),
    HumanMessage(content='¿Cuánto es 10 * 5?'),
    AIMessage(content='50'),
    HumanMessage(content='¿Cuánto es 10 + 3?'),
]

# chat_model.invoke(mensajes)

for tramo in chat_model.stream(mensajes):
    print(tramo.content)
````

Esto es similar a la estructura de mensajes de la API de OpenAI, pero con una sintaxis diferente:

```python
mensajes = [
    {'role': 'user', 'content': 'Cuánto es 1 + 1'},
    {'role': 'assistant', 'content': '2'},
    {'role': 'user', 'content': 'Cuánto es 10 * 5'},
    {'role': 'assistant', 'content': '50'},
    {'role': 'user', 'content': 'Cuánto es 10 + 3'},
]
```

````python
# LangChain v1.0: Messages se importan del namespace simplificado
from langchain.messages import HumanMessage, AIMessage

mensajes = [
    SystemMessage(content='Eres un asistente que resuelve operaciones matemáticas y cuya respuesta es directa.'),
    HumanMessage(content='¿Cuánto es 1 + 1?'),
    AIMessage(content='2'),
    HumanMessage(content='¿Cuánto es 10 * 5?'),
    AIMessage(content='50'),
    HumanMessage(content='¿Cuánto es 10 + 3?'),
]

response = chat_model.invoke(mensajes)
print(response.content)

# for tramo in chat_model.stream(mensajes):
#     print(tramo.content, end='')
````

`ChatModel` es la abstracción **principal** en LangChain v1.0 — recibe mensajes y devuelve mensajes. La antigua clase `LLM` (que recibía y devolvía solo texto plano) es una abstracción **legacy** que se mantiene solo por compatibilidad. La mayoría de proveedores modernos exponen sus modelos exclusivamente como `ChatModel`.

### Patrón moderno: cache por modelo (LangChain v1.0)

El patrón anterior con `set_llm_cache()` activa el cache **globalmente** para toda la aplicación. En v1.0 también puedes asignar un cache **directamente a un modelo específico**, sin afectar a los demás:

```python
chat = init_chat_model("gpt-4.1", cache=InMemoryCache())
```

**Ventajas:**

- ✅ Cada modelo puede tener su propio cache independiente.
- ✅ No contamina el estado global de la aplicación.
- ✅ Útil cuando tienes múltiples modelos y solo quieres cachear algunos.

````python
# Primero limpiamos el cache global para que la prueba sea justa
from langchain_core.globals import set_llm_cache
set_llm_cache(None)

# Creamos un modelo con su propio cache (no global)
from langchain_community.cache import InMemoryCache

chat_con_cache = init_chat_model("gpt-4.1", cache=InMemoryCache())

# Otro modelo SIN cache, para comparar
chat_sin_cache = init_chat_model("gpt-4.1")

print("✅ chat_con_cache tiene cache propio")
print("✅ chat_sin_cache NO tiene cache (independiente)")
````

````python
%%time
# Primera llamada con cache → lenta (consulta a la API)
chat_con_cache.invoke(mensagens)
````

````python
%%time
# Segunda llamada con el MISMO modelo (chat_con_cache) → rapidísima (cache hit)
chat_con_cache.invoke(mensagens)
````

````python
%%time
# La misma pregunta al modelo SIN cache → lenta (NO se beneficia del cache de chat_con_cache)
chat_sin_cache.invoke(mensagens)
````
