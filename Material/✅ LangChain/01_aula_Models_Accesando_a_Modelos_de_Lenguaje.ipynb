## <span style="color:orange">**LLMs**</span>

Modelos de Lenguaje Grande (LLMs, por sus siglas en inglés) son un componente central de LangChain. LangChain no proporciona sus propios LLMs, sino que ofrece una interfaz estándar para interactuar con diferentes LLMs. Específicamente, esta interfaz recibe como entrada una cadena de texto (String) y devuelve una cadena de texto (String).

Existen muchos proveedores de LLMs (OpenAI, Cohere, Hugging Face, etc.) - la clase LLM está diseñada para proporcionar una interfaz estándar para todos ellos.

#### ChatModel con Configuración (LangChain v1.0)

### <span style="color:skyblue">**Llamando al LLM (BATCH)**</span>

```python
pregunta = 'Cuenta una historia breve sobre la travesía de aprender a programar'
respuesta = chat_model.invoke(pregunta)
print(respuesta)
print("-----------------")
print(respuesta.content)
```

### <span style="color:skyblue">**Llamando al LLM con stream**</span>

```python
pregunta = 'Cuenta una historia breve sobre la travesía de aprender a programar'
for tramo in chat_model.stream(pregunta):
    print(tramo.content, end='')
```

### <span style="color:skyblue">**Llamadas simultáneas**</span>

```python
preguntas = [
    '¿Qué es el cielo?',
    '¿Qué es la tierra?',
    '¿Qué son las estrellas?'
]

respuestas = chat_model.batch(preguntas)
#print(respuestas)

# Imprimir solo el contenido en formato legible
for i, respuesta in enumerate(respuestas):
    print(f"Pregunta {i + 1}: {preguntas[i]}")
    print(f"Respuesta: {respuesta.content}\n")
```

Existen 4 tipos principales de mensajes en LangChain v1.0:

- `HumanMessage`: Representa un mensaje del usuario. Generalmente consiste solo en contenido.

- `AIMessage`: Representa un mensaje del modelo. Puede incluir `tool_calls` para llamadas a herramientas y `content_blocks` para acceder a contenido estructurado.

- `SystemMessage`: Representa un mensaje del sistema, que indica al modelo cómo comportarse. Generalmente consiste solo en contenido.

- `ToolMessage`: Representa el resultado de una llamada a herramienta. Incluye un parámetro `tool_call_id` que identifica la llamada que produjo este resultado.

**Nota:** `FunctionMessage` está deprecado y se ha movido a `langchain-classic`. Se recomienda usar `ToolMessage` en su lugar.

**Nuevo en v1.0:** Los mensajes ahora incluyen la propiedad `content_blocks` que proporciona acceso unificado a texto, imágenes, audio, razonamiento y citas del modelo.

### <span style="color:skyblue">**Content Blocks (Nuevo en LangChain v1.0)**</span>

En versiones anteriores de LangChain, la respuesta del modelo se accedía únicamente a través de `respuesta.content` (un string de texto plano). El problema es que los modelos modernos ya no devuelven solo texto: pueden devolver **razonamiento (reasoning)**, **llamadas a herramientas (tool_use)**, **citas de fuentes**, **imágenes**, **audio**, entre otros tipos de contenido.

**LangChain v1.0** introduce la propiedad `content_blocks`, que ofrece acceso **unificado y estructurado** a todos esos tipos de contenido, sin importar el proveedor del modelo (OpenAI, Google, Anthropic, etc.).

Cada bloque tiene un `type` que indica de qué se trata:

- `text` → texto normal de la respuesta
- `reasoning` → cadena de razonamiento del modelo (útil con modelos de razonamiento como Gemini 2.5 o GPT-5)
- `tool_use` → llamada a una herramienta
- `image`, `audio` → contenido multimodal
- `citation` → cita de una fuente

Esto permite procesar la respuesta de forma mucho más rica que con `respuesta.content`.

```python
# Hacemos una pregunta al modelo (Gemini 2.5 Flash devuelve también un bloque de razonamiento)
respuesta = chat_model.invoke("Explica en una frase qué es la fotosíntesis.")

# Recorremos los content_blocks de la respuesta
print(f"Total de bloques: {len(respuesta.content_blocks)}\n")

for i, bloque in enumerate(respuesta.content_blocks):
    print(f"--- Bloque {i + 1} ---")
    print(f"Tipo: {bloque['type']}")
    # Mostramos un fragmento del contenido según su tipo
    if bloque['type'] == 'text':
        print(f"Texto: {bloque['text']}")
    elif bloque['type'] == 'reasoning':
        print(f"Razonamiento: {bloque.get('reasoning', '')[:200]}...")
    else:
        print(f"Contenido: {bloque}")
    print()
```
